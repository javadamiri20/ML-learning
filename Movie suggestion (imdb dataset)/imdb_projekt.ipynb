```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.neighbors import NearestNeighbors
```

```python
df = pd.read_csv('imdb_csv.csv')
df.head(5)
```

```python
df.drop(columns=['Poster_Link', 'Released_Year', 'Certificate',
                 'Runtime', 'IMDB_Rating', 'Overview', 'Meta_score',
                 'No_of_Votes', 'Gross'], inplace=True)
df.head(5)
```

```python
split_Genre = df["Genre"].str.get_dummies(sep=", ")
df = pd.concat((df, split_Genre), axis=1)
df = df.drop(columns='Genre')
```

```python
columns = ['Director','Star1','Star2','Star3','Star4']
df[columns] = df[columns].apply(lambda x: x.str.replace(" ", "_"), axis=1)
```

```python
encoder = LabelEncoder()
col = ['Director','Star1','Star2','Star3','Star4']
for i in col:
    df[i] = encoder.fit_transform(df[i])
```

```python
df
```

```python
def movie(name):
    nearst = NearestNeighbors(n_neighbors=10, algorithm="brute")
    new_film = df[(df['Series_Title'] == name)]
    x_new_film = new_film.drop(columns=['Series_Title'])
    df_new = df.drop(columns=['Series_Title'])
    nearst.fit(df_new)
    sim_film = nearst.kneighbors(x_new_film)
    return df.iloc[sim_film[1][0][1:]]['Series_Title']
```

```python
movie('The Godfather')
```
