```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv('Snappfood_Sentiment_Analysis.csv', sep='\t')
df = df[['comment','label']]
df
```

```python
vectorizer = TfidfVectorizer(ngram_range=(1,3), max_features=50)
text_mosbat = df[df['label'] == 'HAPPY']
vectorized_mosbat = vectorizer.fit_transform(text_mosbat['comment'])
#vectorized_mosbat = vectorizer.transform(text_mosbat['comment'])
```

```python
feature_mosbat = vectorizer.get_feature_names_out(vectorized_mosbat)
feature_mosbat
```

```python
vectorizer2 = TfidfVectorizer(ngram_range=(1,3), max_features=50)
text_manfi = df[df['label'] == 'SAD']
vectorized_manfi = vectorizer2.fit_transform(text_manfi['comment'])
#vectorized_manfi = vectorizer2.transform(text_manfi['comment'])
```

```python
feature_manfi = vectorizer2.get_feature_names_out(vectorized_manfi)
feature_manfi
```
